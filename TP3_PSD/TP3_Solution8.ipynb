```python
# Problem 8. Duplicates and Messy Text
#Step1
from pathlib import Path
import pandas as pd
DATA = Path("Datasets_for_TP3")

reg = pd.read_csv(DATA / "registrations.csv") 
print("Duplicated rows:", reg.duplicated().sum()) 
```

```python
#Step2
reg["name"] = reg["name"].str.strip().str.title() 
reg["city"] = reg["city"].str.strip().str.title() 
print("Duplicated rows:", reg.duplicated().sum()) 
```

```python
#Step3
reg = reg.drop_duplicates().reset_index(drop=True) 
print(reg)
```

```python
#Step4
print("Sum as text  :", reg["fee"].sum()) 
reg["fee"] = reg["fee"].str.replace("$", "").astype(int) 
print("Sum as number:", reg["fee"].sum())
```
