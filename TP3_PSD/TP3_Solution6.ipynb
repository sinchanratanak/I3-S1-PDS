```python
#Problem 6. Creating New Columns
#Step1

from pathlib import Path
import pandas as pd
DATA = Path("Datasets_for_TP3")
import numpy as np 
 
df = pd.read_csv(DATA / "students.csv") 
df["gpa_percent"] = (df["gpa"] / 4.0 * 100).round(1)
```

```python
#Step2
df["grade"] = np.where(df["gpa"] >= 3.7, "A", 
                       np.where(df["gpa"] >= 3.3, "B", "C"))
```

```python
#Step3
def age_group(age): 
    if age < 20: 
        return "teen" 
    elif age <= 21: 
        return "20-21" 
    else: 
        return "22+" 
 
df["age_group"] = df["age"].apply(age_group)
```

```python
#Step4
df["code"] = df["major"] + "-" + df["id"].astype(str)
```

```python
#Step5
print(df[["name", "gpa_percent", "grade", "age_group", "code"]])
```
