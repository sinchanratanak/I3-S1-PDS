```python
# Problem 10. Combining Tables
#Setp1

from pathlib import Path
import pandas as pd
DATA = Path("Datasets_for_TP3")

df = pd.read_csv(DATA / "students.csv") 
scores = pd.read_csv(DATA / "scores.csv")
```

```python
#Step2
merged = pd.merge(df[["id", "name"]], scores, on="id", how="inner") 
merged["total"] = merged["python"] + merged["stats"] 
print(merged.sort_values("total", ascending=False))
```

```python
#Step3
left = pd.merge(df[["id", "name"]], scores, on="id", how="left") 
print("No scores yet:", left[left["python"].isna()]["name"].tolist())
```

```python
#Step4
new = pd.read_csv(DATA / "new_students.csv") 
everyone = pd.concat([df, new], ignore_index=True) 
print("Shape after concat:", everyone.shape) 
print(everyone.tail(3))
```
