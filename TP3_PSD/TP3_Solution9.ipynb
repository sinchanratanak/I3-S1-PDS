```python
# Problem 9. Grouping and Aggregating
#Step1
from pathlib import Path
import pandas as pd
DATA = Path("Datasets_for_TP3")

df = pd.read_csv(DATA / "students.csv") 
print(df.groupby("major")["gpa"].mean().round(2)) 
```

```python
#Step2
summary = df.groupby("major").agg( 
    students=("name", "count"), 
    mean_gpa=("gpa", "mean"), 
    max_gpa=("gpa", "max"), 
    mean_age=("age", "mean"), 
).round(2) 
print(summary)
```

```python
#Step3
print(df.groupby(["major", "gender"])["gpa"].mean().round(2).unstack())
```

```python
#Step4
best_rows = df.groupby("major")["gpa"].idxmax() 
print(df.loc[best_rows, ["major", "name", "gpa"]])
```
