```python
#Step 1
grades = {
    "Dara": [80, 75, 90],
    "Sophea": [95, 88, 92],
    "Vibol": [40, 55, 45],
    "Malis": [70, 85, 78],
    "Kosal": [65, 70, 60],
}
```

```python
#Step 2
average = {}
for name, scores in grades.items():
    average[name] = round(sum(scores) / len(scores), 1)
print("Averages: ", average)
```

```python
#step 3
passed = set()
for name, avg in average.items():
    if avg >= 50:
        print(name, "Pass")
        passed.add(name)
    else:
        print(name, "Fail")
print("Passed: ", sorted(passed))
```

```python
#Step 4
ranking = sorted( average.items(), key=lambda item: item[1], reverse=True) 
for rank, (name, avg) in enumerate(ranking, start=1): 
    print(f"#{rank} {name} ({avg})") 
```
